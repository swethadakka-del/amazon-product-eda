----> Clean the Price Columns

₹1,999
₹5,499
₹39,999

```python
df["discounted_price"] = (
    df["discounted_price"]
    .astype(str)
    .str.replace("₹", "", regex=False)
    .str.replace(",", "", regex=False)
    .str.strip()
)

df["actual_price"] = (
    df["actual_price"]
    .astype(str)
    .str.replace("₹", "", regex=False)
    .str.replace(",", "", regex=False)
    .str.strip()
)

df["discounted_price"] = pd.to_numeric(
    df["discounted_price"],
    errors="coerce"
)

df["actual_price"] = pd.to_numeric(
    df["actual_price"],
    errors="coerce"
)

print("Price columns converted successfully!")
```

----> Clean Discount Percentage

50%
20%
70%

```python
df["discount_percentage"] = (
    df["discount_percentage"]
    .astype(str)
    .str.replace("%", "", regex=False)
    .str.strip()
)

df["discount_percentage"] = pd.to_numeric(
    df["discount_percentage"],
    errors="coerce"
)

print("Discount percentage converted successfully!")
```

----> Clean Rating

```python
df["rating"] = pd.to_numeric(
    df["rating"],
    errors="coerce"
)

print("Rating converted successfully!")
```

----> Clean Rating Count

```python
df["rating_count"] = (
    df["rating_count"]
    .astype(str)
    .str.replace(",", "", regex=False)
    .str.strip()
)

df["rating_count"] = pd.to_numeric(
    df["rating_count"],
    errors="coerce"
)

print("Rating count converted successfully!")
```

----> HISTOGRAM : It tells you how Amazon product prices are distributed.

```python
plt.figure(figsize=(10, 6))

sns.histplot(
    data=df,
    x="discounted_price",
    bins=30,
    kde=True
)

plt.title("Distribution of Discounted Prices")
plt.xlabel("Discounted Price")
plt.ylabel("Number of Products")

plt.show()
```

---> HISTOGRAM FOR MULTIPLE VARIABLES

```python
columns = [
    "discounted_price",
    "actual_price",
    "discount_percentage",
    "rating",
    "rating_count"
]

for column in columns:

    plt.figure(figsize=(10, 6))

    sns.histplot(
        data=df,
        x=column,
        bins=30,
        kde=True
    )

    plt.title(f"Distribution of {column}")
    plt.xlabel(column.replace("_", " ").title())
    plt.ylabel("Frequency")

    plt.show()
```

----> BOXPLOT : Boxplots help identify outliers.

```python
for column in columns:

    plt.figure(figsize=(10, 5))

    sns.boxplot(
        data=df,
        x=column
    )

    plt.title(f"Boxplot of {column}")
    plt.xlabel(column.replace("_", " ").title())

    plt.show()
```

---> KDE[KDE = Kernel Density Estimation] PLOTS : It shows the smooth distribution of a variable.

```python
for column in columns:

    plt.figure(figsize=(10, 6))

    sns.kdeplot(
        data=df,
        x=column,
        fill=True
    )

    plt.title(f"KDE Plot of {column}")
    plt.xlabel(column.replace("_", " ").title())
    plt.ylabel("Density")

    plt.show()
```

---> Bar Chart :
Let's see which categories have the most products.





```python
category_counts = df["category"].value_counts().head(15)

plt.figure(figsize=(12, 7))

sns.barplot(
    x=category_counts.values,
    y=category_counts.index
)

plt.title("Top 15 Product Categories")
plt.xlabel("Number of Products")
plt.ylabel("Category")

plt.show()
```

---->Count Plot

```python
top_categories = df["category"].value_counts().head(15).index

plt.figure(figsize=(12, 7))

sns.countplot(
    data=df[df["category"].isin(top_categories)],
    y="category",
    order=top_categories
)

plt.title("Product Count by Category")
plt.xlabel("Count")
plt.ylabel("Category")

plt.show()
```

----> Pie Chart : Let's visualize the top 10 categories.

```python
category_pie = df["category"].value_counts().head(10)

plt.figure(figsize=(10, 10))

plt.pie(
    category_pie.values,
    labels=category_pie.index,
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Top 10 Product Categories")

plt.show()
```

--->  Scatter Plot : This helps us understand the relationship between the original price and selling price.Actual Price vs Discounted Price BIVARIATE ANALYSIS

```python
plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=df,
    x="actual_price",
    y="discounted_price"
)

plt.title("Actual Price vs Discounted Price")
plt.xlabel("Actual Price")
plt.ylabel("Discounted Price")

plt.show()
```

---> Discount vs Rating

```python
plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=df,
    x="discount_percentage",
    y="rating"
)

plt.title("Discount Percentage vs Rating")
plt.xlabel("Discount Percentage")
plt.ylabel("Rating")

plt.show()
```

----> Pair Plot : This creates multiple plots showing relationships between all selected numerical variables.

```python
pair_columns = [
    "discounted_price",
    "actual_price",
    "discount_percentage",
    "rating"
]

pair_df = df[pair_columns].dropna()

sns.pairplot(
    pair_df,
    diag_kind="kde"
)

plt.show()
```

---> Correlation Matrix

```python
correlation_columns = [
    "discounted_price",
    "actual_price",
    "discount_percentage",
    "rating",
    "rating_count"
]

correlation_matrix = df[correlation_columns].corr()

correlation_matrix
```

---> Heatmap

```python
plt.figure(figsize=(12, 8))

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Heatmap")

plt.show()
```

---> Average Rating by Category
This is another useful bivariate analysis.

```python
category_rating = (
    df.groupby("category")["rating"]
    .mean()
    .sort_values(ascending=False)
    .head(15)
)

plt.figure(figsize=(12, 7))

sns.barplot(
    x=category_rating.values,
    y=category_rating.index
)

plt.title("Average Rating by Product Category")
plt.xlabel("Average Rating")
plt.ylabel("Category")

plt.show()
```

----> Average Discount by Category

```python
category_discount = (
    df.groupby("category")["discount_percentage"]
    .mean()
    .sort_values(ascending=False)
    .head(15)
)

plt.figure(figsize=(12, 7))

sns.barplot(
    x=category_discount.values,
    y=category_discount.index
)

plt.title("Average Discount Percentage by Category")
plt.xlabel("Average Discount (%)")
plt.ylabel("Category")

plt.show()
```
